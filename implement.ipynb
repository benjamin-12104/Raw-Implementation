```python
cv = 5 
times = len(df)//cv
print(times)
for i in range(0, len(df),times):
    train = pd.concat([df["YearsExperience"].iloc[:i], df["YearsExperience"].iloc[i+times:]])
    test = df["YearsExperience"].iloc[i:i+times]
    
    idx = i//times+1 
    print(f"Fold {idx}:")
    print(f"Train: {len(train)}")
    # print(len(train))
    # print()
    print(f"Test: {len(test)}")
    # print(len(test))
    print()
```

```python
cv = 5
fold_size = len(df) // cv

for i in range(cv):
    start = i * fold_size
    end = (i + 1) * fold_size if i < cv - 1 else len(df)

    test = df.iloc[start:end]
    train = pd.concat([df.iloc[:start], df.iloc[end:]])

    print("Fold:", i + 1)
    print(f"Start: {start}, End: {end}")
    print("Train rows:", len(train))
    print("Test rows:", len(test))
    print()
```

```python
# Cosine Similarity 
import numpy as np 

x = [1,2,3]
y = [2,3,4]
sm = 0
a,b = 0,0
for i in range(len(x)):
    sm += x[i]*y[i]
    a += x[i]**2
    b += y[i]**2
# print(sm/np.sqrt(a)*np.sqrt(b))
print(f'Cosine Similarity of {x} and {y} is {round(sm/np.sqrt(a*b),4)}')
```
