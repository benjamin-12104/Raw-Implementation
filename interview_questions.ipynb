```python
# binary search
a = [1,2,3,4,5,6,7,8,9]
target = 10

left = 0
right = len(a)-1
val = False
for i in range(0,len(a)-1):
    mid = left+((right-left)//2)
    if a[mid]==target:
        val = True

    elif a[mid]>target:
        right = mid

    else:
        left = mid
print("Exists" if val else "Not Exists")
```

```python
# perfect number 

num = 6
res = sum([i for i in range(1,num) if num%i==0])
print("perfect number" if res==num else "not a perfect number")
```

```python
# Count of words without using built in methods
string = "the quick brown fox jumps over the lazy dog the quick brown fox runs across the green field"

clean_arr = []
arr = []
start, end = 0,0
i,j = 0,0
while i<len(string):
  start = i
  j = i + 1
  while j<len(string):
    if string[j] == " ":
      end = j
      s = string[start: end]
      arr += [s]
      if s not in clean_arr: clean_arr += [s]
      break
    else:
      j = j + 1
  i = j + 1
    
print(clean_arr)

size = [0]*(len(clean_arr))
print(size)
print()

for i in range(len(clean_arr)):
  for j in range(len(arr)):
    if clean_arr[i]==arr[j]:
      size[i] += 1
print(size)
```

```python
# Second Largest Number - Logic 2 
# arr = [1,10,3,11,6,18,4]
arr = [10, 18, 15, 5]
maxi = float('-inf')
second = float('-inf')
for i in arr:
    if i>maxi:
        second = maxi
        maxi = i
    elif i>second and i!=maxi:
        second = i
print(f"Second Largest: {second}")
```
