```python
# binary search
a = [1,2,3,4,5,6,7,8,9]
target = 10

left = 0
right = len(a)-1
val = False
for i in range(0,len(a)-1):
    mid = left+((right-left)//2)
    if a[mid]==target:
        val = True

    elif a[mid]>target:
        right = mid

    else:
        left = mid
print("Exists" if val else "Not Exists")
```

```python
# Two Sum Logic 1 
nums = [2, 7, 11, 15]
target = 9

for i in range(0,len(nums)):
    for j in range(i+1, len(nums)):
        if nums[i]+nums[j]==target:
            print(f'ids: {i},{j}')
            break
```

```python
# Two Sum Logic 2
nums = [2, 7, 11, 15]
target = 9
seen = {}
for idx, val in enumerate(nums):
    bal = target - val
    if bal in seen.keys():
        print(f"ids: {idx},{seen[bal]}")
        break
    else:
        seen[val] = idx
```

```python
# Rotate an array right by k positions.

# arr:[1, 2, 3, 4, 5, 6, 7]
# 1st rotation: [7, 1, 2, 3, 4, 5, 6]
# 2nd rotation: [6, 7, 1, 2, 3, 4, 5]
# 3rd rotation: [5, 6, 7, 1, 2, 3, 4]

arr = [1, 2, 3, 4, 5, 6, 7]
k = 3
index = 0
for i in range(k):
    last = arr[-1]
    arr = arr[:-1]
    arr += [0]
    j = len(arr)-1
    while j>index:
        arr[j] = arr[j-1]
        j -= 1
    arr[index] = last
    print(arr)
print()
print(f'Final K-{k} Rotated Array: {arr}')
```
